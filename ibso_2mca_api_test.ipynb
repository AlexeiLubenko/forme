# IBSO / 2MCA API test

Тестовый POST-запрос к внутреннему API с XML body.

```python
import requests

url = "http://10.244.227.5:7003/2mca/api"

xml_body = """
<Request>
    <ViewDataGetCancelable
        SessionID=""
        ViewShortName="VW_CRIT_AC_FIN"
        ClassID="AC_FIN"
        Hint="FIRST_ROWS(1)"
        OrderBy="U_1"
        AllowTimestampMillisecounds="true"
        RowsLimit="10000">

        <ClassFilter>
            <ClassInfo ClassID="AC_FIN"/>
        </ClassFilter>

        <UserFilter>
            <AND>
                <SimpleFilter
                    ColumnName="C_1"
                    Operator="LIKE"
                    Value="42%"
                />

                <RoundFilter
                    ColumnName="C_11"
                    Operator="&lt;"
                    Value="0.00"
                    Precision="2"
                />

                <SimpleFilter
                    ColumnName="C_21"
                    Operator="IS NULL"
                />
            </AND>
        </UserFilter>

    </ViewDataGetCancelable>
</Request>
""".strip()

r = requests.post(
    url,
    data=xml_body.encode("utf-8"),
    headers={
        "Content-Type": "application/xml; charset=UTF-8",
        "Accept": "application/xml",
    },
    timeout=30,
)

print("status:", r.status_code)
print("response type:", r.headers.get("Content-Type"))
print(r.text[:10000])
```
